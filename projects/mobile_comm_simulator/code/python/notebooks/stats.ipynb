```python
dict_measurements = {
    'age': [23, 45, 31, 67, 34, 56, 43, 21, 22, 34],
    'height': [1.75, 1.65, 1.72, 1.80, 1.77, 1.90, 1.74, 1.60, 1.70, 1.73],
}
```

```python
df_measurements = pd.DataFrame(dict_measurements)
```

```python
df_measurements
```

## Medidas Resumo

### Medidas de Posição

```python
# mean calculation
mean_age = df_measurements['age'].mean()
mean_age
```

```python
# median calculation
median_age = df_measurements['age'].median()
median_age
```

```python
# mode calculation
mode_age = df_measurements['age'].mode()
mode_age
```

### Medidas de Dispersão

```python
# variance
variance_age = df_measurements['age'].var()
variance_age
```

```python
# standard deviation
std_dev_age = df_measurements['age'].std()
std_dev_age
```

```python
# range
range_age = df_measurements['age'].max() - df_measurements['age'].min()
range_age
```

```python
# Coeficient of variation
cv_age = std_dev_age / mean_age * 100
cv_age
```

### Medidas de Forma

```python
# Age Asymmetry
skewness_age = df_measurements['age'].skew()
skewness_age
```

```python
# Height Asymmetry
skewness_height = df_measurements['height'].skew()
skewness_height
```

```python
# Age Kurtosis
kurtosis_age = df_measurements['age'].kurtosis()
kurtosis_age
```

```python
# Height Kurtosis
kurtosis_height = df_measurements['height'].kurtosis()
kurtosis_height
```

```python
df_measurements.describe()
```

### Correlação entre Variáveis  

```python
# Pearson correlation
pearson_correlation = df_measurements.corr(method='pearson')
pearson_correlation
```

```python
# Spearman correlation
spearman_correlation = df_measurements.corr(method='spearman')
spearman_correlation
```

```python

# Correlation between age and height
correlation_age_height = df_measurements['age'].corr(df_measurements['height'])
correlation_age_height
```
